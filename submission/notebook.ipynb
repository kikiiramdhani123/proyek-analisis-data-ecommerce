## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memperoleh informasi yang mendalam: **S**pecific, **M**easurable, **A**ction-oriented, **R**elevant, dan **T**ime-bound.

Dataset ini berisi transaksi marketplace Olist di Brasil (September 2016 – Oktober 2018). Karena data pada 2016 dan September–Oktober 2018 sangat sedikit (tidak mewakili bulan penuh), seluruh analisis dibatasi pada **pesanan berstatus `delivered` dengan tanggal pembelian 1 Januari 2017 – 31 Agustus 2018** (20 bulan penuh). **Revenue** didefinisikan sebagai total `price` barang pada `order_items` (tanpa ongkos kirim), dalam mata uang Real Brasil (BRL).

- **Pertanyaan 1:** Bagaimana tren jumlah pesanan dan total revenue per bulan selama periode Januari 2017 – Agustus 2018, pada bulan apa puncak penjualan terjadi, dan berapa persen pertumbuhan *Month-over-Month* (MoM) pada bulan puncak tersebut?
  - *Specific*: tren bulanan pesanan & revenue · *Measurable*: jumlah pesanan, revenue (BRL), % MoM · *Action-oriented*: menentukan bulan untuk kampanye besar dan penambahan kapasitas stok/logistik · *Relevant*: pertumbuhan penjualan adalah tujuan utama marketplace · *Time-bound*: Jan 2017 – Agu 2018.
- **Pertanyaan 2:** Kategori produk apa saja yang masuk 10 besar dan 10 terbawah berdasarkan total revenue selama Januari 2017 – Agustus 2018, dan berapa persen kontribusi gabungan 10 kategori teratas terhadap total revenue?
  - *Specific*: revenue per kategori produk · *Measurable*: revenue (BRL) dan % kontribusi · *Action-oriented*: prioritas anggaran promosi, akuisisi seller, dan evaluasi kategori berkinerja rendah · *Relevant*: alokasi sumber daya pemasaran · *Time-bound*: Jan 2017 – Agu 2018.
- **Pertanyaan 3:** Berapa selisih rata-rata review score antara pesanan yang terlambat dan pesanan yang tepat waktu (dibanding estimasi pengiriman) selama Januari 2017 – Agustus 2018, dan 10 negara bagian (*state*) pelanggan mana yang memiliki persentase keterlambatan tertinggi?
  - *Specific*: keterlambatan vs kepuasan pelanggan, per state · *Measurable*: rata-rata review score (1–5) dan % pesanan terlambat · *Action-oriented*: menentukan wilayah prioritas perbaikan logistik dan akurasi estimasi pengiriman · *Relevant*: kepuasan pelanggan memengaruhi pembelian ulang · *Time-bound*: Jan 2017 – Agu 2018.

### Cleaning Data

#### Memperbaiki tipe data datetime (orders, order_items, order_reviews)

```python
datetime_cols = ['order_purchase_timestamp', 'order_approved_at', 'order_delivered_carrier_date',
                 'order_delivered_customer_date', 'order_estimated_delivery_date']
for col in datetime_cols:
    orders_df[col] = pd.to_datetime(orders_df[col])

order_items_df['shipping_limit_date'] = pd.to_datetime(order_items_df['shipping_limit_date'])
for col in ['review_creation_date', 'review_answer_timestamp']:
    reviews_df[col] = pd.to_datetime(reviews_df[col])

orders_df[datetime_cols].dtypes
```

#### Memfilter pesanan delivered pada periode analisis dan membuang data inkonsisten (orders)

```python
orders_clean_df = orders_df[
    (orders_df['order_status'] == 'delivered') &
    (orders_df['order_purchase_timestamp'] >= START_DATE) &
    (orders_df['order_purchase_timestamp'] < pd.Timestamp(END_DATE) + pd.Timedelta(days=1))
].dropna(subset=['order_delivered_customer_date']).copy()

# Fitur pengiriman: lama pengiriman (hari) dan status terlambat dibanding estimasi
orders_clean_df['delivery_days'] = (
    orders_clean_df['order_delivered_customer_date'] - orders_clean_df['order_purchase_timestamp']
).dt.total_seconds() / 86400
orders_clean_df['is_late'] = (
    orders_clean_df['order_delivered_customer_date'].dt.normalize() > orders_clean_df['order_estimated_delivery_date']
)

print('Jumlah pesanan sebelum:', len(orders_df))
print('Jumlah pesanan sesudah:', len(orders_clean_df))
print('Rentang tanggal:', orders_clean_df['order_purchase_timestamp'].min(), 's.d.', orders_clean_df['order_purchase_timestamp'].max())
orders_clean_df.isna().sum()
```

#### Memperbaiki missing value dan kategori tanpa terjemahan (products)

```python
products_df['product_category_name'] = products_df['product_category_name'].fillna('unknown')
for col in ['product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']:
    products_df[col] = products_df[col].fillna(products_df[col].median())
products_df = products_df.drop(columns=['product_name_lenght', 'product_description_lenght', 'product_photos_qty'])

extra_translation = pd.DataFrame({
    'product_category_name': ['portateis_cozinha_e_preparadores_de_alimentos', 'pc_gamer', 'unknown'],
    'product_category_name_english': ['kitchen_portables_and_food_preparers', 'pc_gamer', 'unknown'],
})
category_translation_df = pd.concat([category_translation_df, extra_translation], ignore_index=True)

products_df = products_df.merge(category_translation_df, on='product_category_name', how='left')
print('Missing value products:\n', products_df.isna().sum(), sep='')
```

#### Memperbaiki invalid dan inaccurate value (order_payments)

```python
payments_df = payments_df[payments_df['payment_type'] != 'not_defined'].copy()
payments_df.loc[payments_df['payment_installments'] == 0, 'payment_installments'] = 1

print(payments_df['payment_type'].value_counts())
print('\nCicilan bernilai 0:', (payments_df['payment_installments'] == 0).sum())
```

#### Menghapus duplikat review per pesanan (order_reviews)

```python
reviews_df = (reviews_df.sort_values('review_answer_timestamp')
                        .drop_duplicates(subset='order_id', keep='last')
                        .drop(columns=['review_comment_title', 'review_comment_message']))
print('Duplikat order_id:', reviews_df['order_id'].duplicated().sum())
print('Jumlah review:', len(reviews_df))
```

**Insight:**
- Kolom tanggal pada tabel orders, order_items, dan order_reviews sudah bertipe `datetime64`.
- Setelah difilter (status delivered, periode Januari 2017 – Agustus 2018, dan membuang 8 data inkonsisten), tersisa 96.203 pesanan dari total 99.441. Ditambahkan dua fitur baru: `delivery_days` (lama pengiriman) dan `is_late` (terlambat dibanding estimasi).
- Kategori kosong diisi `unknown` dan dua terjemahan ditambahkan, sehingga tabel products tidak lagi memiliki missing value.
- Transaksi `not_defined` dihapus dan cicilan 0 diubah menjadi 1.
- Review sudah dideduplikasi menjadi 98.673 baris, satu review per pesanan.

### Explore tren pesanan dan revenue bulanan

```python
monthly_df = (all_df.set_index('order_purchase_timestamp')
                    .resample('MS')
                    .agg(order_count=('order_id', 'nunique'), revenue=('price', 'sum'))
                    .reset_index())
monthly_df['month'] = monthly_df['order_purchase_timestamp'].dt.strftime('%Y-%m')
monthly_df['order_mom_pct'] = monthly_df['order_count'].pct_change() * 100
monthly_df['revenue_mom_pct'] = monthly_df['revenue'].pct_change() * 100
monthly_df[['month', 'order_count', 'revenue', 'order_mom_pct', 'revenue_mom_pct']]
```

```python
peak = monthly_df.loc[monthly_df['revenue'].idxmax()]
print(f"Bulan puncak revenue : {peak['month']} | revenue {peak['revenue']:,.2f} BRL | {peak['order_count']:,} pesanan")
print(f"Pertumbuhan MoM puncak: revenue {peak['revenue_mom_pct']:.1f}% | pesanan {peak['order_mom_pct']:.1f}%")

yearly = monthly_df.assign(year=monthly_df['order_purchase_timestamp'].dt.year)
jan_aug = yearly[yearly['order_purchase_timestamp'].dt.month <= 8].groupby('year')[['order_count', 'revenue']].sum()
jan_aug['revenue_growth_pct'] = jan_aug['revenue'].pct_change() * 100
print('\nPerbandingan Jan-Agu 2017 vs Jan-Agu 2018:')
jan_aug
```

```python
# Hari dengan pesanan terbanyak pada November 2017
nov_df = all_df[all_df['order_purchase_timestamp'].dt.to_period('M') == '2017-11']
nov_daily = (nov_df.groupby(nov_df['order_purchase_timestamp'].dt.date)['order_id'].nunique()
                   .sort_values(ascending=False))
nov_daily.head(5)
```

**Insight:**
- Pesanan tumbuh dari 750 pesanan (111,8 ribu BRL) pada Januari 2017 menjadi kisaran 6.000–7.000 pesanan per bulan sepanjang 2018.
- Puncak terjadi pada **November 2017**: 7.288 pesanan dengan revenue 987.648 BRL, naik **62,8% (pesanan) dan 52,4% (revenue) MoM**. Hari dengan pesanan terbanyak adalah 24 November 2017 (1.147 pesanan), yaitu hari **Black Friday**.
- Revenue Desember 2017 turun 26,5% setelah Black Friday, lalu pulih pada Januari 2018.
- Revenue Januari–Agustus 2018 sebesar 7,22 juta BRL, naik 141,1% dibanding periode yang sama pada 2017 (2,99 juta BRL). Namun, sejak Mei 2018 revenue cenderung stagnan bahkan menurun (Juni 2018 turun 12,5% MoM).

### Explore revenue per kategori produk

```python
category_df = (all_df.groupby('product_category')
                     .agg(revenue=('price', 'sum'), order_count=('order_id', 'nunique'), items_sold=('order_item_id', 'count'))
                     .sort_values('revenue', ascending=False)
                     .reset_index())
category_df['share_pct'] = category_df['revenue'] / category_df['revenue'].sum() * 100
category_df['cum_share_pct'] = category_df['share_pct'].cumsum()
print('Jumlah kategori:', len(category_df))
category_df.head(10)
```

```python
category_df.tail(10)
```

```python
top10_share = category_df.head(10)['share_pct'].sum()
n_80 = (category_df['cum_share_pct'] < 80).sum() + 1
print(f'Kontribusi 10 kategori teratas: {top10_share:.1f}% dari total revenue')
print(f'Jumlah kategori yang menyumbang 80% revenue: {n_80} dari {len(category_df)}')
```

**Insight:**
- Terdapat 74 kategori produk. Kategori dengan revenue tertinggi adalah `health_beauty` (1,23 juta BRL; 9,3%), diikuti `watches_gifts` (1,16 juta BRL; 8,8%) dan `bed_bath_table` (1,02 juta BRL; 7,8%).

## Analisis Lanjutan (Opsional)

### RFM Analysis

**Tujuan:** RFM (*Recency, Frequency, Monetary*) mengelompokkan pelanggan berdasarkan perilaku transaksinya agar tim pemasaran bisa memberi perlakuan berbeda untuk tiap kelompok (retensi, reaktivasi, atau program loyalitas).

- **Recency**: jumlah hari sejak pembelian terakhir pelanggan hingga tanggal acuan (1 hari setelah transaksi terakhir di dataset). Semakin kecil semakin baik.
- **Frequency**: jumlah pesanan unik per pelanggan (`customer_unique_id`).
- **Monetary**: total revenue (harga barang) dari pelanggan tersebut.

**Teknik skoring:** Recency dan Monetary diberi skor 1–5 berdasarkan kuintil (5 = terbaik). Karena sebagian besar pelanggan hanya membeli sekali, kuintil tidak bisa dipakai untuk Frequency, sehingga skornya ditentukan manual: 1 = 1 pesanan, 2 = 2 pesanan, 3 = ≥3 pesanan. Segmen kemudian ditentukan dengan aturan (*rule-based clustering*) dari kombinasi skor tersebut.

```python
reference_date = all_df['order_purchase_timestamp'].max() + pd.Timedelta(days=1)
rfm_df = all_df.groupby('customer_unique_id').agg(
    last_purchase=('order_purchase_timestamp', 'max'),
    frequency=('order_id', 'nunique'),
    monetary=('price', 'sum'),
).reset_index()
rfm_df['recency'] = (reference_date - rfm_df['last_purchase']).dt.days

rfm_df['R_score'] = pd.qcut(rfm_df['recency'], 5, labels=[5, 4, 3, 2, 1]).astype(int)
rfm_df['M_score'] = pd.qcut(rfm_df['monetary'], 5, labels=[1, 2, 3, 4, 5]).astype(int)
rfm_df['F_score'] = np.select([rfm_df['frequency'] >= 3, rfm_df['frequency'] == 2], [3, 2], default=1)

def segment(row):
    if row['F_score'] >= 2 and row['R_score'] >= 4:
        return 'Champions'
    if row['F_score'] >= 2:
        return 'Loyal (pembeli ulang)'
    if row['R_score'] >= 4 and row['M_score'] >= 4:
        return 'Pelanggan baru bernilai tinggi'
    if row['R_score'] >= 4:
        return 'Pelanggan baru'
    if row['R_score'] == 3:
        return 'Perlu perhatian'
    if row['M_score'] >= 4:
        return 'Berisiko hilang (bernilai tinggi)'
    return 'Hibernasi'

rfm_df['segment'] = rfm_df.apply(segment, axis=1)
print('Tanggal acuan:', reference_date.date())
print(f"Pelanggan dengan 1 pesanan: {(rfm_df['frequency'] == 1).mean():.1%}")
rfm_df[['recency', 'frequency', 'monetary']].describe()
```

```python
segment_df = (rfm_df.groupby('segment')
                    .agg(customers=('customer_unique_id', 'count'), avg_recency=('recency', 'mean'),
                         avg_frequency=('frequency', 'mean'), avg_monetary=('monetary', 'mean'),
                         total_monetary=('monetary', 'sum'))
                    .sort_values('customers', ascending=False))
segment_df['customer_pct'] = segment_df['customers'] / segment_df['customers'].sum() * 100
segment_df['revenue_pct'] = segment_df['total_monetary'] / segment_df['total_monetary'].sum() * 100
segment_df
```

```python
plot_seg = segment_df.sort_values('customers')
highlight_segments = ['Berisiko hilang (bernilai tinggi)', 'Champions', 'Loyal (pembeli ulang)']
fig, axes = plt.subplots(1, 2, figsize=(15, 5.5), sharey=True)
colors = [ACCENT if s.startswith('Berisiko') else HIGHLIGHT if s in highlight_segments else MUTED for s in plot_seg.index]

axes[0].barh(plot_seg.index, plot_seg['customers'], color=colors)
for i, (v, p) in enumerate(zip(plot_seg['customers'], plot_seg['customer_pct'])):
    axes[0].text(v + 200, i, f'{v:,} ({p:.1f}%)', va='center', fontsize=9, color=TEXT)
axes[0].set_xlim(0, plot_seg['customers'].max() * 1.3)
axes[0].set_xlabel('Jumlah pelanggan')
axes[0].set_title('Pembeli ulang (biru) hanya 3%, segmen bernilai tinggi\nyang berisiko hilang (oranye) 14.5%', loc='left')

axes[1].barh(plot_seg.index, plot_seg['avg_monetary'], color=colors)
for i, v in enumerate(plot_seg['avg_monetary']):
    axes[1].text(v + 5, i, f'{v:,.0f} BRL', va='center', fontsize=9, color=TEXT)
axes[1].set_xlim(0, plot_seg['avg_monetary'].max() * 1.3)
axes[1].set_xlabel('Rata-rata monetary per pelanggan (BRL)')
axes[1].set_title('Rata-rata nilai belanja per segmen', loc='left')
for ax in axes:
    ax.grid(axis='x', color='#e8e7e3', linewidth=0.8)
plt.tight_layout()
plt.show()
```
